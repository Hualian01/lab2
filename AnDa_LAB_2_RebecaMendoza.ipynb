## Genere un histograma y diagrama de densidad para la columna 9 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine


# Cargar el conjunto de datos de vinos
wine_data = load_wine()
wine_df = pd.DataFrame(wine_data.data, columns=wine_data.feature_names)

# Crear un diagrama de densidad para la columna 9
sns.kdeplot(wine_df['proanthocyanins'], color='orange')
plt.title('Densidad de Proanthocyanins en Vinos', fontsize=14)
plt.xlabel('Proanthocyanins')
plt.ylabel('Densidad')
plt.show()

# Añadir la columna target al DataFrame
wine_df['target'] = wine_data.target

# Seleccionar la columna 'proanthocyanins' del DataFrame
proanthocyanins_col = wine_df['proanthocyanins']

# Configurar el estilo de Seaborn
sns.set(style="whitegrid")

# Crear la figura y los ejes para el histograma y el diagrama de densidad
fig, axs = plt.subplots(2, 1, figsize=(10, 8))

# Histograma
sns.histplot(data=proanthocyanins_col, kde=False, color='skyblue', ax=axs[0])
axs[0].set_title('Histograma de proanthocyanins')

# Diagrama de densidad
sns.kdeplot(data=proanthocyanins_col, color='orange', shade=True, ax=axs[1])
axs[1].set_title('Diagrama de densidad de proanthocyanins')

# Mostrar la figura
plt.tight_layout()
plt.show()

```

## Aplique el Escalado Z a la columna 2 y grafique el antes y despues

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets

# Cargar el dataset Wine
X, y = datasets.load_wine(return_X_y=True)
wine_data = X[:, None, 2]

# Calcular la media y la desviación estándar para el escalado Z
avg = np.average(wine_data)
std = np.std(wine_data)

# Aplicar el escalado Z a la columna 2 del dataset de vinos
z_scaled = (wine_data - avg) / std

# Crear la figura y los ejes para el gráfico
fig, axs = plt.subplots(2, 1, sharex=True, tight_layout=True)

# Graficar el histograma antes del escalado Z (en color verde)
axs[0].hist(wine_data, color='green')
axs[0].set_title('Original')
axs[0].set_yticks([0, 10, 20, 30, 40, 50,60,70])

# Graficar el histograma después del escalado Z (en color rojo)
axs[1].hist(z_scaled, color='red')
axs[1].set_title('Escalado Z')
axs[1].set_yticks([0, 10, 20, 30, 40, 50, 60, 70])

# Mostrar la figura
plt.show()

```

## Cambie en el dataset los valores de la columna MONTH por una abreviatura de los respectivos meses 

```python
import pandas as pd

# Read the CSV file into a DataFrame
datos = pd.read_csv("Warehouse_and_Retail_Sales.csv")

# Create a mapping dictionary for the month numbers and abbreviated names
month_mapping = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun",
    7: "Jul",
    8: "Aug",
    9: "Sep",
    10: "Oct",
    11: "Nov",
    12: "Dec"
}

# Apply the mapping to the MONTH column
datos['MONTH'] = datos['MONTH'].map(month_mapping)

# Print the updated DataFrame
datos.head()
```

## Presente un gráfico de barras un crosstab entre el mes y el año

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo CSV en un DataFrame
#datos = pd.read_csv("Warehouse_and_Retail_Sales.csv")

# Crear una tabla de contingencia entre las columnas MONTH y YEAR
crosstab = pd.crosstab(datos['MONTH'], datos['YEAR'])

# Definir los colores que deseas utilizar
colors = ['blue', 'green', 'red', 'orange', 'purple', 'brown']

# Graficar la tabla de contingencia como un gráfico de barras con colores personalizados
crosstab.plot(kind='bar', rot=0, color=colors)
plt.xlabel('Mes')
plt.ylabel('Cantidad')
plt.title('Tabla de Contingencia entre Mes y Año')
plt.legend(title='Año')
plt.show()
```
